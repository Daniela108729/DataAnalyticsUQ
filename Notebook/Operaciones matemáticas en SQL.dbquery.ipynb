```sql
CREATE TABLE IF NOT EXISTS workspace.electiva.indicadores_paises (
    pais_id INT,
    region_economica STRING,
    pais STRING,
    pib_billones_usd DOUBLE,          -- Producto Interno Bruto en billones de USD
    tasa_inflacion_pct DECIMAL(5,2),  -- Porcentaje de inflación anual
    tasa_desempleo_pct STRING,        -- Almacenado como STRING para practicar CAST
    subsidios_millones STRING,        -- Contiene NULLs cuando no se registran subsidios
    poblacion_millones INT
) USING DELTA;
```
