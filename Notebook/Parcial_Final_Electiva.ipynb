## Paso 1. crear las tablas delta a partir de los archivos csv

### Tabla customers

```sql
CREATE OR REPLACE TABLE workspace.electiva.customers
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS CustomerID,
    _c1 AS FirstName,
    _c2 AS LastName,
    _c3 AS FullName
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/customers.csv`
WHERE _c0 <> 'CustomerID';
```

### Tabla employees

```sql
CREATE OR REPLACE TABLE workspace.electiva.employees
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS EmployeeID,
    CASE
        WHEN _c1 = 'NULL' THEN NULL
        ELSE CAST(_c1 AS INT)
    END AS ManagerID,
    _c2 AS FirstName,
    _c3 AS LastName,
    _c4 AS FullName,
    _c5 AS JobTitle,
    CAST(_c6 AS INT) AS OrganizationLevel,
    _c7 AS MaritalStatus,
    _c8 AS Gender,
    CASE WHEN _c9 = 'NULL' THEN NULL ELSE _c9 END AS Territory,
    CASE WHEN _c10 = 'NULL' THEN NULL ELSE _c10 END AS Country,
    CASE WHEN _c11 = 'NULL' THEN NULL ELSE _c11 END AS `Group`
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/employees.csv`
WHERE _c0 <> 'EmployeeID';
```

### Tabla orders

```sql
CREATE OR REPLACE TABLE workspace.electiva.orders
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS SalesOrderID,
    CAST(_c1 AS INT) AS SalesOrderDetailID,
    TO_DATE(_c2, 'M/d/yyyy') AS OrderDate,
    TO_DATE(_c3, 'M/d/yyyy') AS DueDate,
    TO_DATE(_c4, 'M/d/yyyy') AS ShipDate,
    CAST(_c5 AS INT) AS EmployeeID,
    CAST(_c6 AS INT) AS CustomerID,
    CAST(_c7 AS DOUBLE) AS SubTotal,
    CAST(_c8 AS DOUBLE) AS TaxAmt,
    CAST(_c9 AS DOUBLE) AS Freight,
    CAST(_c10 AS DOUBLE) AS TotalDue,
    CAST(_c11 AS INT) AS ProductID,
    CAST(_c12 AS INT) AS OrderQty,
    CAST(_c13 AS DOUBLE) AS UnitPrice,
    CAST(_c14 AS DOUBLE) AS UnitPriceDiscount,
    CAST(_c15 AS DOUBLE) AS LineTotal
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/orders.csv`
WHERE _c0 <> 'SalesOrderID';
```

### Tabla product_categories

```sql
CREATE OR REPLACE TABLE workspace.electiva.product_categories
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS CategoryID,
    _c1 AS CategoryName
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/productcategories.csv`
WHERE _c0 <> 'CategoryID';
```

### Tabla Product_sub_categories

```sql
CREATE OR REPLACE TABLE workspace.electiva.product_sub_categories
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS SubCategoryID,
    CAST(_c1 AS INT) AS CategoryID,
    _c2 AS SubCategoryName
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/productsubcategories.csv`
WHERE _c0 <> 'SubCategoryID';
```

### Tabla products

```sql
CREATE OR REPLACE TABLE workspace.electiva.products
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS ProductID,
    _c1 AS ProductNumber,
    _c2 AS ProductName,
    _c3 AS ModelName,
    CAST(_c4 AS INT) AS MakeFlag,
    CAST(_c5 AS DOUBLE) AS StandardCost,
    CAST(_c6 AS DOUBLE) AS ListPrice,
    CAST(_c7 AS INT) AS SubCategoryID
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/products.csv`
WHERE _c0 <> 'ProductID';
```

### Tabla vendors

```sql
CREATE OR REPLACE TABLE workspace.electiva.vendor
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS VendorID,
    _c1 AS VendorName,
    _c2 AS AccountNumber,
    CAST(_c3 AS INT) AS CreditRating,
    CAST(_c4 AS INT) AS ActiveFlag
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/vendors.csv`
WHERE _c0 <> 'VendorID';
```

### Tabla vendors_product

```sql
CREATE OR REPLACE TABLE workspace.electiva.vendor_product
USING DELTA
AS
SELECT
    CAST(_c0 AS INT) AS ProductID,
    CAST(_c1 AS INT) AS VendorID
FROM csv.`/Volumes/workspace/electiva/archivosparcial_csv/vendorproduct.csv`
WHERE _c0 <> 'ProductID';
```

```sql
CREATE OR REPLACE VIEW workspace.electiva.orders_august_2011 AS

WITH ordenes_agosto AS (
    SELECT *
    FROM workspace.electiva.orders
    WHERE OrderDate >= DATE '2011-08-01'
      AND OrderDate < DATE '2011-09-01'
)

SELECT
    p.ProductName AS productname,
    p.ModelName AS modelname,
    pu.user_name AS user_name,
    o.SalesOrderID AS salesorderid,
    pc.CategoryName AS categoryname,
    o.OrderDate AS orderdate,
    o.LineTotal AS linetotal,

    CASE
        WHEN pc.CategoryName ILIKE 'Bikes' THEN 0.10
        WHEN pc.CategoryName ILIKE 'Clothing' THEN 0.15
        WHEN pc.CategoryName ILIKE 'Components' THEN 0.08
        WHEN pc.CategoryName ILIKE 'Accessories' THEN 0.20
        ELSE 0
    END AS `descuento_mitad_año`,

    o.LineTotal -
    (
        o.LineTotal *
        CASE
            WHEN pc.CategoryName ILIKE 'Bikes' THEN 0.10
            WHEN pc.CategoryName ILIKE 'Clothing' THEN 0.15
            WHEN pc.CategoryName ILIKE 'Components' THEN 0.08
            WHEN pc.CategoryName ILIKE 'Accessories' THEN 0.20
            ELSE 0
        END
    ) AS total_descuentos_aplicados

FROM ordenes_agosto o

LEFT JOIN workspace.electiva.products p
    ON o.ProductID = p.ProductID

LEFT JOIN workspace.electiva.product_sub_categories ps
    ON p.SubCategoryID = ps.SubCategoryID

LEFT JOIN workspace.electiva.product_categories pc
    ON ps.CategoryID = pc.CategoryID

LEFT JOIN workspace.electiva.page_user pu
    ON o.CustomerID = pu.customerid;
```

```sql
SELECT *
FROM workspace.electiva.orders_august_2011;
```

```sql
SELECT
    categoryname,
    `descuento_mitad_año`,
    COUNT(*) AS total_ordenes
FROM workspace.electiva.orders_august_2011
GROUP BY
    categoryname,
    `descuento_mitad_año`
ORDER BY categoryname;
```

## Punto 4

```sql
CREATE OR REPLACE MATERIALIZED VIEW workspace.electiva.customer_class
AS
SELECT
    c.CustomerID AS customerid,
    c.FullName AS customer_fullname,
    pu.user_name AS user_name,

    COUNT(DISTINCT o.SalesOrderID) AS total_orders,

    CASE
        WHEN COUNT(DISTINCT o.SalesOrderID) = 3 THEN 'Oro'
        WHEN COUNT(DISTINCT o.SalesOrderID) = 2 THEN 'Plata'
        ELSE 'Ocasional'
    END AS user_level

FROM workspace.electiva.customers c

LEFT JOIN workspace.electiva.page_user pu
    ON c.CustomerID = pu.customerid

LEFT JOIN workspace.electiva.orders o
    ON c.CustomerID = o.CustomerID
    AND o.OrderDate >= DATE '2011-01-01'
    AND o.OrderDate < DATE '2012-01-01'

GROUP BY
    c.CustomerID,
    c.FullName,
    pu.user_name;
```

```sql
SELECT *
FROM workspace.electiva.customer_class
ORDER BY total_orders DESC;
```

```sql
SELECT
    user_level,
    COUNT(*) AS total_clientes
FROM workspace.electiva.customer_class
GROUP BY user_level
ORDER BY total_clientes DESC;
```
